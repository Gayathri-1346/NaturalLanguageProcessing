```python
grammar={
    ("Det","N"):"NP",
    ("v","NP"):"VP",
    ("NP","VP"):"S"
}
```

```python
lexicon={
    "the":"Det",
    "cat":"N",
    "dog":"N",
    "chased":"V",
}
```

```python
sentence="the cat chased the dog".split()
stack=[]
buffer=sentence.copy()
print("Initial Buffer:",buffer)
print()
while buffer:
    word=buffer.pop(0)
    stack.append(lexicon[word])
    print("SHIFT:",word)
    print("stack:",stack)
    print()
    reduced=True
    while reduced:
        reduced=False
        for rule,lhs in grammar.items():
            if tuple(stack[-len(rule):])==rule:
                stack=stack[:-len(rule)]
                stack.append(lhs)
                print("REDUCE:",rule,"->",lhs)
                print("stack:",stack)
                print()
                reduce=True
                break
if stack==["S"]:
    print("sentence Accepted")
else:
    print("sentence Rejected")
```

```python
import nltk
from nltk import CFG 
from nltk.parse.shiftreduce import ShiftReduceParser
```

```python
grammar=CFG.fromstring("""
S -> NP VP
NP -> Det N
VP -> V NP
Det -> 'the'
N -> 'cat'|'dog'
V -> 'chased' """)
print(grammar)
```

```python
parser=ShiftReduceParser(grammar)
```

```python
sentence="the cat chased the dog".split()
print("Input Sentence:",sentence)
print("\n parse Tree\n")
for tree in parser.parse(sentence):
    print(tree)
    tree.pretty_print()
```
