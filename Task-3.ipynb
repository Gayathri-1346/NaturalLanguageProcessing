```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import nltk
```

```python
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
text=input("Enter text:")
words=word_tokenize(text)
print(words)
```

```python
topics={"Health care":['patient','doctor','hospital','medicine','healthcare'],
    "Technology": ["computer", "software", "internet", "AI", "programming"],
    "Education": ["school", "student", "teacher", "learning", "education"],
    "Sports": [ "football", "cricket", "player", "match", "tournament"],
    "Finance": ["bank", "money", "investment", "stock", "finance"],
    "Politics": ["government", "election", "policy", "minister", "parliament"],
    "Business": ["company", "market", "customer", "sales", "profit"]
    }
```

```python
segmented_topics={"Healthcare":[],
                  "Finance":[],
                  "Sports":[]}
```

```python
for sentence in text:
    sentence_lower=sentence.lower()
    for topic, keywords in topics.items():
        if any(keyword in sentence_lower
               for keyword in keywords):
            segmented_topics[topic].append(sentence)

for topic,content in segmented_topics.items():
    print("\n",topic)
    print("-----")
    for sentence in content:
        print(sentence)
```

```python
text=input("Enter text:").lower()
found="unknown"
for topic, words in topics.items():
    for word in words:
        if word in text:
            found=topic
            break
print("Topic:",found)
```
