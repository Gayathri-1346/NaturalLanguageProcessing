```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
```

```python
nltk.download('wordnet')
```

```python
nltk.download('punkt_tab')
```

```python
text="""The turkey hit the wall and slid down, leaving a trail of grease.
My mum stood at one end of the table, nostrils flared, still wearing her apron, her sleeves rolled up showing the bruises on her arms, the knife shook in her hand.
My dad sat, his eyes distant. Suddenly, he banged both fists on the table. I snatched the carving fork and stood next to mum. My sister cowered behind us.
He rose from his seat, swayed a little and looked like he was about to speak. Instead, he turned and left. It was the best Christmas ever."""
```

```python
#removal of stop words

from stopwords import removal_sw
text=text.lower()
text=removal_sw(text)
```

```python
#Tokenization
tokens=word_tokenize(text)
print(tokens)
```

```python
# Stemming
res_stem=[]
stemmer=PorterStemmer()
for i in tokens:
    res_stem.append(stemmer.stem(i))
print(res_stem)
```

```python
#lemmatizing
lemma=WordNetLemmatizer()
res_lemma=[]
for i in tokens:
    res_lemma.append(lemma.lemmatize(i,pos="v"))
print(res_lemma)
```

```python
import pandas as pd
pd.DataFrame({
    "Original_text":tokens,
    "Stemming": res_stem,
    "Leammatizing": res_lemma
})
```
