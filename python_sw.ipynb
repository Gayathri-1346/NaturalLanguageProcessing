```python
stop_word=["if","then","a","is","are","and","to"]
sent1="Today is a beautiful day to explore new ideas and improve your skills"
res=""
word=sent1.split()
print(word)
for i in range(len(word)):
    if word[i] not in stop_word:
        res+=word[i]+""
print("After removing stopwords:",res)
```

```python
f2=open("file2.txt","r")
x=f2.read()
f2.close()
words=x.split()
ans=""
print("Before removing stopwords:",words,"\n")
for i in range(len(words)):
    if words[i] not in stop_word:
        ans+=words[i]+""
print("After removing stopwords:",ans)
```

```python
f3=open("file3.txt","w")
f3.write(ans)
f3.close()
```
