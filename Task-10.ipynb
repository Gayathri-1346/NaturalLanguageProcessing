```python
import re

import nltk


# Download required NLTK resources

nltk.download("stopwords")


from nltk.corpus import stopwords

from nltk.tokenize import word_tokenize


# 1. INPUT TEXT


text = input("Enter text: ")


print("\nOriginal Text:")

print(text)

# 2. LOWERCASE


text = text.lower()


# 3. REMOVE PUNCTUATION AND SPECIAL CHARACTERS


text = re.sub(r"[^a-zA-Z0-9\s]", "", text)


# 4. TOKENIZATION


tokens = text.split()



# 5. REMOVE STOPWORDS


stop_words = set(stopwords.words("english"))


tokens = [

word for word in tokens

if word not in stop_words

]

# 6. PREPARE CLEANED TEXT



cleaned_text = " ".join(tokens)


print("\nCleaned Text:")

print(cleaned_text)

# 7. DISPLAY TOKENS



print("\nTokens:")

print(tokens)


# 8. PREPARE FOR MODEL INPUT

model_input = tokens


print("\nModel Input:")

print(model_input)
```

```python
import re

import numpy as np

import tensorflow as tf

from tensorflow.keras.layers import TextVectorization

# 1. INPUT TEXT

text = input("Enter text: ")


print("\nOriginal Text:")

print(text)

# 2. BASIC CLEANING

# Convert to lowercase

text = text.lower()


# Remove punctuation

text = re.sub(r"[^\w\s]", "", text)


print("\nCleaned Text:")

print(text)

# 3. CREATE TEXT VECTORIZATION LAYER

vectorizer = TextVectorization(

max_tokens=1000,

output_mode="int",

output_sequence_length=20

)


# Adapt the vectorizer to the input text

vectorizer.adapt([text])

# 4. CONVERT TEXT INTO NUMBERS


vectorized_text = vectorizer([text])


print("\nVocabulary:")

print(vectorizer.get_vocabulary())


print("\nModel Input:")

print(vectorized_text.numpy())
```
