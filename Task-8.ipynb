```python
import spacy
nlp=spacy.load("en_core_web_sm")
text=input("Enter Text:")
doc=nlp(text)
print("Named Entities:")
for ent in doc.ents:
    print(ent.text,"->",ent.label_)
print("---BIO TAGGING---")
print(f"{'Token':20}{'BIO Tag'}")
print("-"*30)
for token in doc:
    if token.ent_iob_ == "B":
        bio_tag="B-"+token.ent_type_
    elif token.ent_iob_ == "I":
        bio_tag="I-"+token.ent_type_
    else:
        bio_tag="O"
    print(f"{token.text:20}{bio_tag}")
```

```python
from transformers import pipeline
ner=pipeline("token-classification",model="d4data/biomedical-ner-all")
text=input("Enter medical text")
res=ner(text)
print("\n"+"="*60)
print("BIO TAGGGING")
print("="*60)
print(f"{'Token':30}{'BIO TAG'}")
print("-"*60)
for item in res:
    token=item['word']
    tag=item['entity']
    print(f"{token:30}{tag}")
print("\n+"+"="*60)
print("MEDICAL NER")
print("="*60)
entities=[]
curr_entity=""
curr_type=""
for item in res:
    token=item["word"]
    tag=item["entity"]
    if tag=='O':
        if curr_entity:
            entities.append((curr_entity,curr_type))
            curr_entity=""
            curr_type=""
        continue
    if "-" in tag:
        prefix, entity_type=tag.split("-",1)
    else:
        prefix="B"
        entity_type=tag
    if prefix=="B":
        if curr_entity:
            entities.append((curr_entity,curr_type))
        curr_entity=token
        curr_type=entity_type
    elif prefix=="I":
        if curr_entity:
            curr_entity+=""+token
        else:
            curr_entity=token
        curr_type=entity_type
if curr_entity:
    entities.append((curr_entity,curr_type))
for entity, entity_type in entities:
    print(f"{entity:40}->{entity_type}")
```
