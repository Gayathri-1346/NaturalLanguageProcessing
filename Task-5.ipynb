```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('stop_words')
nltk.download('punkt_tab')
```

```python
a="Artificial Narrow Intelligence is the only form of AI that currently exists. ANI models are designed to perform a single, specific task, such as identifying images, engaging in chat, or filtering emails."
```

```python
words=word_tokenize(a)
print(words)
```

```python
stop_words=set(stopwords.words('english'))
print(stop_words)
```

```python
filtered_words=[word for word in words if word.lower() not in stop_words]
print(filtered_words)
```

```python
print("Original words:\n",words)
print("\n After stopwords removal:\n",filtered_words)
```

#### Using string.punctutaion

```python
filter_str_words=[word for word in words if word.lower() not in string.punctuation]
print(filter_str_words)
```

#### Custom stopwords

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))
custom_stopwords = {'form', 'AI', 'currently','.',',',':'}
stop_words.update(custom_stopwords)
print(stop_words)
print("\n")
filter_words=[word for word in words if word.lower() not in stop_words]
print(filter_words)
```
