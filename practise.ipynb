```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
```

```python
nltk.download('punkt_tab')
```

```python
f1=open('text.txt','r')
```

```python
a=f1.read()
print(a)
```

```python
words=word_tokenize(a)
```

```python
words
```

```python
words_stem=[]
words_lemma=[]
```

```python
stemm=PorterStemmer()
for i in words:
    words_stem.append(stemm.stem(i))
words_stem
```

```python
lemma=WordNetLemmatizer()
for i in words:
    words_lemma.append(lemma.lemmatize(i,pos='v'))
words_lemma
```

```python
import pandas as pd
```

```python
df=pd.DataFrame({
    "Original":words,
    "Stemming":words_stem,
    "Lemmatization":words_lemma})
df
```

```python
import nltk
from nltk.corpus import wordnet as wn
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def word_sense_disambiguation(sentence, target_word):
    senses=wn.synsets(target_word)
    if not senses:
        print("Word not found in WordNet")
        return
    definitions=[]
    for sense in senses:
        definition=sense.definition()
        examples="".join(sense.examples())
        definitions.append(definition+""+examples)
    documents=[sentence]+definitions
    vectorizer=TfidfVectorizer(stop_words='english')
    vectors=vectorizer.fit_transform(documents)
    similarities=cosine_similarity(vectors[0:1],vectors[1:])[0]
    best_index=similarities.argmax()
    best_sense=senses[best_index]
    print("\nSentence:",sentence)
    print("Word:",target_word)
    print("\nMost Likely sense:")
    print("Sense:",best_sense.name())
    print("Meaning:",best_sense.definition())
    print("Similarity:",round(similarities[best_index],3))
    if best_sense.examples():
        print("Example:",best_sense.examples()[0])
sentence=input("Enter a Sentence:")
word=input("Enter the word whose sense you want to find:")
word_sense_disambiguation(sentence,word)
```
