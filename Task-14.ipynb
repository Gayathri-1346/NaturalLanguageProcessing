```python
# ------------------------------------------------------------
# 1. Import Libraries
# ------------------------------------------------------------

import numpy as np
import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

print("Libraries imported successfully.")
print("TensorFlow version:", tf.__version__)
```

```python
# ------------------------------------------------------------
# 2. Set Random Seeds
# ------------------------------------------------------------

np.random.seed(42)
tf.random.set_seed(42)

print("Random seeds set successfully.")
```

```python
# ------------------------------------------------------------
# 3. Create a Small Text Dataset
# ------------------------------------------------------------

text = """
machine learning is a branch of artificial intelligence.
machine learning allows computers to learn from data.
deep learning uses neural networks to solve complex problems.
neural networks are useful for image and text processing.
natural language processing helps computers understand language.
artificial intelligence is used in many real world applications.
"""

# Convert text to lowercase
text = text.lower()

print("Text dataset created successfully.")
print("\nSample text:")
print(text[:300])
```

```python
# ------------------------------------------------------------
# 4. Create Character Vocabulary
# ------------------------------------------------------------

characters = sorted(list(set(text)))

char_to_int = {
    char: i
    for i, char in enumerate(characters)
}

int_to_char = {
    i: char
    for i, char in enumerate(characters)
}

vocab_size = len(characters)

print("Vocabulary size:", vocab_size)
print("\nCharacters in vocabulary:")
print(characters)
```

```python
# ------------------------------------------------------------
# 5. Convert Characters to Integers
# ------------------------------------------------------------

encoded_text = np.array([
    char_to_int[char]
    for char in text
], dtype=np.int32)

print("Encoded text created successfully.")
print("\nFirst 50 encoded characters:")
print(encoded_text[:50])
```

```python
# ------------------------------------------------------------
# 6. Create Input Sequences and Target Characters
# ------------------------------------------------------------

sequence_length = 40

X = []
y = []

for i in range(
    len(encoded_text) - sequence_length
):

    sequence = encoded_text[
        i:i + sequence_length
    ]

    target = encoded_text[
        i + sequence_length
    ]

    X.append(sequence)
    y.append(target)

X = np.array(X, dtype=np.int32)
y = np.array(y, dtype=np.int32)

print("Input sequences created successfully.")
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
# ------------------------------------------------------------
# 7. Display a Sample Sequence
# ------------------------------------------------------------

sample_sequence = X[0]

sample_text = "".join(
    int_to_char[index]
    for index in sample_sequence
)

target_character = int_to_char[y[0]]

print("Input sequence:")
print(sample_text)

print("\nTarget character:")
print(target_character)
```

```python
# ------------------------------------------------------------
# 8. Build the LSTM Model
# ------------------------------------------------------------

model = Sequential([
    
    Embedding(
        input_dim=vocab_size,
        output_dim=64
    ),

    LSTM(
        128
    ),

    Dense(
        vocab_size,
        activation="softmax"
    )
])

print("LSTM model created successfully.")
```

```python
# ------------------------------------------------------------
# 9. Compile the Model
# ------------------------------------------------------------

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("Model compiled successfully.")
```

```python
# ------------------------------------------------------------
# 10. Display Model Architecture
# ------------------------------------------------------------

model.summary()
```

```python
# ------------------------------------------------------------
# 11. Train the LSTM Model
# ------------------------------------------------------------

history = model.fit(
    X,
    y,
    epochs=30,
    batch_size=32,
    verbose=1
)

print("\nTraining completed successfully.")
```

```python
# ------------------------------------------------------------
# 12. Evaluate the Model
# ------------------------------------------------------------

loss, accuracy = model.evaluate(
    X,
    y,
    verbose=0
)

print("\n====================================")
print("MODEL PERFORMANCE")
print("====================================")

print("Loss    :", round(loss, 4))
print("Accuracy:", round(accuracy * 100, 2), "%")
```

```python
# ------------------------------------------------------------
# 13. Function to Generate Text
# ------------------------------------------------------------

def generate_text(
    model,
    seed_text,
    number_of_characters=100
):

    generated_text = seed_text.lower()

    for _ in range(number_of_characters):

        # Take the last 40 characters
        input_text = generated_text[
            -sequence_length:
        ]

        # If input is shorter than 40 characters,
        # add spaces at the beginning
        if len(input_text) < sequence_length:

            input_text = (
                " " * (
                    sequence_length -
                    len(input_text)
                )
                + input_text
            )

        # Convert characters to integers
        input_sequence = np.array([
            char_to_int.get(
                char,
                char_to_int[" "]
            )
            for char in input_text
        ], dtype=np.int32)

        # Reshape for the model
        input_sequence = input_sequence.reshape(
            1,
            sequence_length
        )

        # Predict the next character
        prediction = model.predict(
            input_sequence,
            verbose=0
        )

        # Select character with highest probability
        predicted_index = np.argmax(
            prediction[0]
        )

        predicted_character = int_to_char[
            predicted_index
        ]

        # Add predicted character
        generated_text += predicted_character

    return generated_text
```

```python
# ------------------------------------------------------------
# 14. Generate New Text
# ------------------------------------------------------------

seed_text = "machine learning"

generated_text = generate_text(
    model,
    seed_text,
    number_of_characters=100
)

print("\n====================================")
print("GENERATED TEXT")
print("====================================")

print(generated_text)
```

```python
# ------------------------------------------------------------
# 15. Generate Another Text Sample
# ------------------------------------------------------------

seed_text_2 = "artificial intelligence"

generated_text_2 = generate_text(
    model,
    seed_text_2,
    number_of_characters=100
)

print("\n====================================")
print("SECOND GENERATED TEXT")
print("====================================")

print(generated_text_2)
```

```python
# ------------------------------------------------------------
# 16. Final Result
# ------------------------------------------------------------

print("\n")
print("==============================================")
print("     CHARACTER-LEVEL LSTM LANGUAGE MODEL")
print("==============================================")

print(
    "Vocabulary Size:",
    vocab_size
)

print(
    "Sequence Length:",
    sequence_length
)

print(
    "Training Accuracy:",
    round(accuracy * 100, 2),
    "%"
)

print("\nModel Architecture:")
print("Character Input")
print("      ↓")
print("Embedding Layer")
print("      ↓")
print("LSTM Layer")
print("      ↓")
print("Dense + Softmax")
print("      ↓")
print("Next Character Prediction")

print("\nCharacter-level language model completed successfully.")
```
