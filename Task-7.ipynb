```python
import nltk
from nltk.corpus import wordnet as wn
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
def word_sense_disambiguation(sentence, target_word):
    senses=wn.synsets(target_word)
    if not senses:
        print("Word not found in WordNet")
        return
    definitions=[]
    for sense in senses:
        definition=sense.definition()
        examples="".join(sense.examples())
        definitions.append(definition+""+examples)
    documents=[sentence]+definitions
    vectorizer=TfidfVectorizer(stop_words='english')
    vectors=vectorizer.fit_transform(documents)
    similarities=cosine_similarity(vectors[0:1],vectors[1:])[0]
    best_index=similarities.argmax()
    best_sense=senses[best_index]
    print("\nSenttence:",sentence)
    print("Word:",target_word)
    print("\nMost Likely sense:")
    print("Sense:",best_sense.name())
    print("Meaning:",best_sense.definition())
    print("Similarity:",round(similarities[best_index],3))
    if best_sense.examples():
        print("Example:",best_sense.examples()[0])
sentence=input("Enter a Sentence:")
word=input("Enter the word whose sense you want to find:")
word_sense_disambiguation(sentence,word)
```
