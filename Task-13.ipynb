```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,classification_report
```

```python
reviews = [
    "This movie was absolutely fantastic. I loved every minute of it.",
    "The acting was brilliant and the story was very engaging.",
    "The movie was boring and the plot was difficult to follow.",
    "I was disappointed by the weak storyline and poor acting.",
    "An amazing film with great performances and beautiful visuals."
]
sentiments = [
    1,
    1,
    0,
    0,
    1
]
```

```python
vectorizer=CountVectorizer()
X=vectorizer.fit_transform(reviews)
y=sentiments
print("Vocabulary:")
print(vectorizer.get_feature_names_out())
print("\nBag of Words Matrix:")
print(X.toarray())
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42,stratify=y)
model=LogisticRegression()
model.fit(X_train,y_train)
y_pred=model.predict(X_test)
accuracy=accuracy_score(y_test,y_pred)
print("\nAccuracy:",accuracy)
print("\nClassification Report:")
print(classification_report(y_test,y_pred))
new_reviews = [
    "The movie had an amazing story and the actors gave outstanding performances.",
    "I enjoyed the film from beginning to end. It was entertaining and exciting."
]
new_X=vectorizer.transform(new_reviews)
predictions=model.predict(new_X)
print("\nSentiment Prediction:")
for review,prediction in zip(new_reviews,predictions):
    sentiment="Positive" if prediction==1 else "Negative"
    print(review,"->",sentiment)
```
